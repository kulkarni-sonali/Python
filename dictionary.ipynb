```python
# LOOPING IN DICTIONARIES 
# It can be done in 3 ways - 1)Key - Value pair, 2) Keys , 2.a) keys by shorter way, 3) values

courses={'DAA': 'CS', 'AOA': 'ME', 'SVY' :'CE'}

# ITERATE OVER KEY- VALUE PAIRS

for K,V in courses.items():
    print(K,V)

print(".........................")

# ITERATE OVER KEY

for K in courses.keys():
    print(K,V)
    print(K)

print(".........................")

# ITERATE OVER KEY ( shorter way)

for K in courses :
    print(K)
    print(K,V)

print(".........................")

# ITERATE OVER VALUE

for V in courses.values() :
    print(V)
    print(V,K)


print(".........................")
```

```python
# CLONING A DICTIONARY
# Clone or create a copy of dictionary 
# having 4 methods to do so
# 1. copy() , 2. dict() , 3. Dictionary unpacking, 4. Copy Module

# 1. copy() method: creates shallow copy of parent dictionary
dict1={'x':10, 'y':20, 'z':30}
print(dict1)
print()

print("..........................")

dict2=dict1.copy() # Formula for copy() method
print(dict1)
print(dict2)

print("..........................")

dict2['a']=40
print(dict1)
print(dict2) # dict1 remains unchanged
```

```python
#CLONING A DICTIONARY
# 2.dict() constructor method
dict3={'a':1, 'b':2, 'c':3}

dict4=dict(dict3) # formula for dict() method

print(dict3)
print(dict4)
print("..........................")
dict4['d']=4
print(dict3)
print(dict4)# dict3 remains unchanged
```

```python
#CLONING A DICTIONARY
# 3. dictionay unpacking
dict5={'a':1, 'b':2, 'c':3}

dict6={**dict5} # formula for dictionary unpacking
print(dict5)
print(dict6)

print("..........................")
dict6['d']=4
print(dict5)
print(dict6) # dict5 remains unchanged
```

```python
# DICTIONARY METHODS 

c={'CS101':'CPP', 'CS102':'DS', 'CS201':'OOP'}
d={'ME126':'HPE', 'ME102':'TOM', 'ME234':'AEM'}


print(c.get('CS101')) # accessing value at given key using get() method
print(c.get('CS101','Absent'))
print(c.get('CS999','Absent')) # accessing value at given key using get() method with default value
print("...........1...............")

c.update(d) # merging two dictionaries
print(c)
print(c.update(d)) # update() method returns None
print(d.update(c)) # update() method returns None
print("..........2................")

print(d.pop('ME102')) # removing key-value pair at given key using pop() method    
print(".........3.................") 

print(c.popitem()) # removing last inserted key-value pair using popitem() method
print(".........4.................")
# How come c gives last key of d ? because c was updated with d in previous step
# how come d gives last key of c ? because d was updated with c in previous step

c.clear() # clearing entire dictionary
print(c)
d.clear() # clearing entire dictionary
print(d)
print(".........5.................")
```
