```python


def Square(My_List):
    """ Finding square of items in the list"""
    Squares=[]
    for i in My_List:
        Squares.append(i**2)
        return squares
lst = [1,2,3,4,5]
print(f"The List is :{lst}")
print(f"The square of items in list is:{Squares(lst)}")
```
