```python
def Square(My_List):
    """ Finding square of items in the list"""
    squares=[]
    for i in My_List:
        squares.append(i**2)
    return squares

lst1 = [1,2,3,4,5]
print(f"The List is :{lst1}")
print(f"The square of items in list is:{Square(lst1)}")
print(Square(lst1))
```

```python
# 2 KEYWORD ARGUMENTS 
# Order specification in arguments is not necessary
def Funct(num1,num2=40):
    print("Number 1 is :" , num1)
    print("Number 2 is :", num2)

# 1 Calling a function without using keyword argument

print("PASSING VALUES WITHOUT KEYWORD ARGUMENTS")
Funct(100) # when no second (or first, either-or) value given that time will take existing value from main function
Funct(100,200)# when both values are defined under keyword then automatically initial value from main fucntion will be cancelled or ignored

# 2 Calling a function by using keyword argument

print("\nPASSING VALUES WITH KEYWORD ARGUMENTS")
Funct(num2=90, num1=20)
Funct(num1=20, num2=90)

# Any order will give same output
    
```

```python
# Exponents / power/ liek square or cube etc.
print(pow(2,3)) # i.e. cube of 2 | or 2 raised to the power of 3 | 2**3 | 2^3

print (pow(y=3, x=2)) # gives typo error bcz The built-in pow() function is defined with positional-only parameters
# that means we must pass arguments by position, not by keyword name.
```

```python
# 3.REQUITRED ARGUMENTS
# required arguments| order specific arguments| number of parameters = number of arguments
def Funct(num1,num2):
    print("Number 1 is:", num1)
    print ("Number 2 is:", num2)
Funct(100,40)
Funct(40,100)
# Here when we change the position or order of the arguments then output also changes which is not the same in KEYWORD ARGUMENTS

Funct(30) #TypeError: Funct() missing 1 required positional argument: 'num2'

Funct(30,50,90) #TypeError: Funct() takes 2 positional arguments but 3 were given
```

```python
# 4 VARIABLE -LENGTH ARGUMENTS
# Example 1
def Funct (*args):
    print("Numbers are :" , args)

Funct(10,25,78)
Funct(17,78,96,48,56,47)
Funct(12)


# Example 2 -Calculating Sum of Numbers
def Sum (*args):
    Total=0
    for val in args:
        Total+=val
    print("Sum:", Total)

Sum()
Sum(10)
Sum(10,20)
Sum(10,20,30)
```

```python
# Multiple arguments with keywords to a single starred vaiable. .... it will give an ERROR

def Funct(*args):
    print("Numbers are :", args)


Funct(10,25,78)
Funct ( num1 = 10, num2 = 25, num3 = 78) # TypeError: Funct() got an unexpected keyword argument 'num1'
```

```python
# **kwargs ----- Accessing the values in the form of KEYWORD ARGUMENTS

def Funct(**kwargs):
    print("Numbers are:", kwargs)
Funct(num1=10, num2=25, num3 =78)
```

```python
def Vals(n1, *s):
    print(n1)
    print(s)
Vals(10)
Vals(10,1,2,3,4,5)
Vals() # TypeError: Vals() missing 1 required positional argument: 'n1'
```

```python
# APPLYING *args  ON THE FUNCTION

def Avg_Num(*num):
    sum=0; length =len(num)

    for i in num:
        sum+=i
    print(f"The Average of Numbers : {sum/length}")

Avg_Num(3,4)
Avg_Num(1,2,3,4)
```

```python
# APPLYING * kwargs on the function

def Avg_Num(**nums):
    sum=0; length =len(nums)

    for key, value in nums.items():
        sum+=value
    print(f"The Average of Numbers : {sum/length}")

Avg_Num(a1=3, a2=4)
Avg_Num(x1=1, x2=2, x3=3, x4=4)
```
