# 1D Array -Creating a 1-D Array
An array with only single row of elements is called 1-D array. 

```python
#Let us try to create a 1-D array from a list which contains numbers as well as strings

array1=np.array([5,-7.4,'a',7.2])
array1

# Observe that since there is a string value in the list, all integer and float values have been promoted to string, while converting the list to array.
# Note: U32 means Unicode-32 data type.
```

Creating a 2-D Array

```python
array2=np.array([[2.4,3],
                 [4.9,7],
                 [0,-1]])
array2

# Observe that the integers 3, 7, 0 and -1 have been promoted to floats.
```

Attributes of NumPy Array

1)    .ndim:  - 
Gives the number of dimensions of the array as an integer value. Arrays can be 1-D, 2-D or n-D.


```python
print(array1.ndim)
print(array2.ndim)
```

2.    shape - It gives the sequence of integers indicating the size of the array for each dimension.



```python
print(array.shape)
print(array1.shape)
print(array2.shape)
```

3)  .size: It gives the total number of elements of the array. This is equal to the product of the elements of shape.

```python
print(array.size)
print(array1.size)
print(array2.size)
```

4)   .dtype:     is the data type of the elements of the array. 
All the elements of an array are of same data type. 
Common data types are int32, int64, float32, float64, U32, etc.

```python
print(array.dtype)
print(array1.dtype)
print(array2.dtype)
```

5)     .itemsize:    It specifies the size in bytes of each element of the array. 
Data type int32 and float32 means each element of the array occupies 32 bits in memory. 
8 bits form a byte. Thus, an array of elements of type int32 has itemsize 32/8=4 bytes. 
Likewise, int64/float64 means each item has itemsize 64/8=8 bytes.

```python
print(array.itemsize)
print(array1.itemsize)
print(array2.itemsize)
```

Other Ways of Creating NumPy Arrays

We can specify data type (integer, float, etc.) while creating array using dtype as an argument to array(). This will convert the data automatically to the mentioned type. In the following example, nested list of integers are passed to the array function. Since data type has been declared as float, the integers are converted to floating point numbers.


```python
array3=np.array([[1,2],[3,4]], dtype=float)
array3

array4p=np.array([3,4],[8,9], dtype=str)
print(arrayp)
```

2) We can create an array with all elements initialised to 0 using the function zeros(). By default, the data type of the array created by zeros() is float. The following code will create an array with 3 rows and 4 columns with each element set to 0.

```python
array4=np.zeros((3,2))
array4

arrayt=np.zeros((3,5))
arrayt
```

3) We can create an array with all elements initialised to 1 using the function ones(). By default, the data type of the array created by ones() is float. The following code will create an array with 3 rows and 2 columns.

```python
array5=np.ones((3,2)) # first number 3 denotes column while second number 2 denotes rows
array5

# Note- as (dtype=int ) instructions avoided after (3,2) therefore output is in float
```

```python
array5=np.ones((3,2), dtype=int) # first number 3 denotes column while second number 2 denotes rows
array5

# Note- as (dtype=int) instruction ahs been given so now output is in  intiger form
```
