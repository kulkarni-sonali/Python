```python
#MATHEMATICAL SET OPERATIONS
# Union, Intersection, Difference, Symmetric Difference

engineers = {'Vijay', 'Ajay', 'Sanjay', 'Ravi','Sujay', 'Dinesh'}
managers = {'Aditya','Sanjay'}
print("Engineers:", engineers)
print("Managers:", managers)   
print(engineers.union(managers))  # Union
print(engineers | managers)  # Union
print(engineers.intersection(managers))  # Intersection
print(engineers & managers)  # Intersection
print(engineers.difference(managers))  # Difference
print(engineers - managers)  # Difference
print(managers.difference(engineers))  # Difference
print(managers - engineers)  # Difference
print(engineers.symmetric_difference(managers))  # Symmetric Difference
print(engineers ^ managers)  # Symmetric Difference
```
