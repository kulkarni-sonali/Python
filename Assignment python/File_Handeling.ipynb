```python
#  Write numbers 1–10, seperate even and odd


even_file = open(r"C:\Users\Sonali\Desktop\even.txt","a") 
odd_file = open(r"C:\Users\Sonali\Desktop\odd.txt","a") 

f = open(r"C:\Users\Sonali\Desktop\number.txt","r") 

lines = f.readlines()

print(lines)

for line in lines: 
    line = line.strip()
    num = int(line)

    if num % 2 == 0: 
        even_file.write(str(num)+"\n") 
    else: 
        odd_file.write(str(num)+"\n") 


even_file.close()
odd_file.close()
```
