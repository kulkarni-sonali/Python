```python
# Difference between lambda function and UDF (User defined function)
# 1

# User Defined Function
def Reciprocal(num):
    return 1/num

# Lambda function definition
Lambda_Reciprocal = lambda num : 1/num

# Taking input from user
val=int(input("Enter the value to find out reciprocal: "))

# Using the function defined by def keyword
print("\n Usinf UDF:", Reciprocal (val))

# Using teh functiin defind by lambda keyword
print("\n Using lambda function: ", Lambda_Reciprocal(val))
```

```python
# filter(function):

#filter()function is used to filter values from the given sequence based on some conditon
#function argument is responsible to perform CONDITIONAL CHECK SEQUENCE , it can be list,tuple or string.
#syntax:
#       filter(function, sequence)

```

```python
# PROGRAM TO FILTER ONLY EVEN NUMBERS FROM THE LIST BY USING THE FILTER()FUNCTION

""" 
1. WITHOUT USING THE LAMBDA FUNCTION  """

def isEven(x):
    if  x%2 == 0:
        return True
    else:
        return False
    
l=[0,5,10,15,20,25,30]
l1=list(filter(isEven,l))
print(l1)

print("\n----------------------------------")

""" 2. WITH USING THE LAMBDA FUNCTION """

isEven=lambda x: x%2==0

l=[0,5,10,15,20,25,30]
l1=list(filter(isEven,l))

print(l1)
```

```python
# MAP(FUNCTION):
# map(function):
#THE FUNCTION CAN BE APPLIED TO EACH ELEMENT OF THE SEQUENCE AND GENERATE A NW SEQUENCE
#SYNTAX:
#   MAP(FUNCTION,SEQUENCE)
```

```python
# WITHOUT USING THE LAMBDA FUNCTION

def DoubleIt(x):
    return x*2

l=[1,2,3,4,5]
l1 = list(map(DoubleIt,l))
print(l1)



# WITH USING THE LAMBDA FUNCTION
l=[1,2,3,4,5]
l1=list(map(lambda x : x*2,l))

print(l1)

```

```python
#syntax:
#   map(lambdax,y: x*y, l1,l2)
# x is from l1 and y is from l2


#EXAMPLE 1

l=[1,2,3,4,5]
l1=list(map(lambda x : x*x,l))
print(l1)
```

```python
# EXAMPLE 2

l1=[1,2,3,4,5]
l2=[2,3,4,5,6]

l3 = list(map(lambda x,y: x*y, l1,l2))
print(l3)
```

```python
# reduce () function:
# it reduces the sequence of elements into a single element by applying the specified function
# present in the functools module and hence we should write an import statement
# syntax: reduce(function,sequence)

from functools import*
l=[10,20,30,40,50]
result = reduce(lambda x,y: x+y,l)
print(result)
```
