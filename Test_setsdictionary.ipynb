```python
courses={
    "Math":[85,67,90],
    "English":[91,95,93],
    "Science":[165,72,891],
    "History":[82,85,88]
}
averages={}
for course,score in courses.items():
    average=sum(score)/len(score)
    averages[course]=average
print(max(averages))
```
